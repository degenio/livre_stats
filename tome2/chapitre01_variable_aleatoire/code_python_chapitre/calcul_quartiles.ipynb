# Code de calcul de la moyenne dans différentes situations.

```python
import numpy as np
import pandas as pd

# Données exemple (notes triées)
notes = np.array([28, 32, 35, 40, 42, 45, 48, 50, 52, 55, 
                  57, 60, 62, 65, 68, 70, 72, 75, 78, 82, 
                  85, 88, 90, 94, 100])
 
# Calcul des quartiles avec numpy (méthode 'linear' par défaut)
print("Quartiles avec numpy (interpolation linéaire) :")
print(f"Q1 = {np.percentile(notes, 25):.2f}")
print(f"Q2 (médiane) = {np.percentile(notes, 50):.2f}")
print(f"Q3 = {np.percentile(notes, 75):.2f}")
print(f"IQR = {np.percentile(notes, 75) - np.percentile(notes, 25):.2f}")

# Avec pandas
s = pd.Series(notes)
print(f"\nAvec pandas : Q1={s.quantile(0.25):.2f}, Q2={s.quantile(0.5):.2f}, Q3={s.quantile(0.75):.2f}")

# Détection des outliers avec la règle 1.5×IQR
Q1 = np.percentile(notes, 25)
Q3 = np.percentile(notes, 75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
print(f"\nDétection des outliers (règle 1.5×IQR) :")
print(f"Bornes : [{lower_bound:.2f}, {upper_bound:.2f}]")
outliers = notes[(notes < lower_bound) | (notes > upper_bound)]
print(f"Valeurs aberrantes : {outliers if len(outliers) > 0 else 'Aucune'}")
```
