# Boxplot

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

# Données exemple (notes de physique)
notes = np.array([78, 82, 85, 88, 90, 92, 95, 98, 100, 102, 105, 
                  100, 102, 75, 78, 60, 54, 40, 85, 50])

# 1. Boxplot simple vertical
plt.figure(figsize=(6, 8))
sns.boxplot(y=notes, width=0.3, color='skyblue', showfliers=True)
plt.title('Distribution des notes de physique')
plt.ylabel('Note (/100)')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
 
# 2. Boxplot horizontal (pour libellés longs)
plt.figure(figsize=(8, 3))
sns.boxplot(x=notes, width=0.3, color='lightgreen', showfliers=True)
plt.title('Distribution des notes (orientation horizontale)')
plt.xlabel('Note (/100)')
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()

# 3. Boxplot groupé par catégorie
donnees_groupes = pd.DataFrame({
    'note': np.concatenate([notes, notes + np.random.normal(0, 5, len(notes))]),
    'groupe': ['Groupe A']*len(notes) + ['Groupe B']*len(notes)
})
plt.figure(figsize=(8, 6))
sns.boxplot(data=donnees_groupes, x='groupe', y='note', palette='Set2', showfliers=True)
plt.title('Comparaison des notes entre deux groupes')
plt.ylabel('Note (/100)')
plt.xlabel('Groupe')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

# 4. Boxplot « notched » pour comparer des médianes
plt.figure(figsize=(8, 6))
sns.boxplot(data=donnees_groupes, x='groupe', y='note', palette='Set2', notch=True)
plt.title('Comparaison des médianes avec intervalles de confiance (encoches)')
plt.ylabel('Note (/100)')
plt.xlabel('Groupe')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

# 5. Boxplot avec points jitterés (pour petits échantillons)
plt.figure(figsize=(6, 8))
sns.boxplot(y=notes, width=0.3, color='lightcoral', showfliers=False)
sns.stripplot(y=notes, color='black', size=4, jitter=0.15, alpha=0.6)
plt.title('Boxplot avec observations individuelles (jitter)')
plt.ylabel('Note (/100)')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```
