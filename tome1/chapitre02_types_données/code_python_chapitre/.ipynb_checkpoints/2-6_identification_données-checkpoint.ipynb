## Identification pratique des données

```python
import pandas as pd
def identifier_type_donnees(serie):
    """
    Fonction d'aide pour identifier le type de données
    """
    # Vérifier si numérique
    if pd.api.types.is_numeric_dtype(serie):
        # Vérifier si entier (potentiellement discret)
        if serie.dtype in ['int64', 'int32']:
            if serie.min() >= 0:  # Pas de valeurs négatives
                print("Possiblement : Rapport (discret)")
            else:
                print("Possiblement : Intervalle (discret)")
        else:
            print("Possiblement : Rapport ou Intervalle (continu)")
    else:
        # Données catégorielles
        valeurs_uniques = serie.nunique()
        if valeurs_uniques <= 10:  # Seuil arbitraire
            print(f"Catégorielle avec {valeurs_uniques} catégories")
            print("À déterminer : Nominal ou Ordinal selon le contexte")
        else:
            print("Possiblement : Nominal avec beaucoup de catégories")

```

```python
# Exemple d'utilisation
data_test = pd.DataFrame({
    'couleur': ['rouge', 'bleu', 'vert', 'rouge'],
    'satisfaction': [1, 2, 3, 2],
    'temperature': [15.5, 18.2, 22.1, 19.8],
    'age': [25, 30, 35, 28]
})

for col in data_test.columns:
    print(f"\nColonne '{col}':")
    identifier_type_donnees(data_test[col]) 
```
