## Pré-traitement de données

```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, StandardScaler
import pandas as pd

# Exemple de pré-traitement selon le type
df_exemple = pd.DataFrame({
    'couleur': ['rouge', 'bleu', 'vert', 'rouge'],  # Nominal
    'satisfaction': ['Faible', 'Moyen', 'Élevé', 'Moyen'],  # Ordinal
    'age': [25, 30, 35, 28]  # Rapport
})

# Encodage nominal (one-hot)
couleur_encoded = pd.get_dummies(df_exemple['couleur'], prefix='couleur')

# Encodage ordinal
satisfaction_mapping = {'Faible': 1, 'Moyen': 2, 'Élevé': 3}
df_exemple['satisfaction_encoded'] = df_exemple['satisfaction'].map(satisfaction_mapping)
   
# Standardisation des données de rapport
scaler = StandardScaler()
df_exemple['age_scaled'] = scaler.fit_transform(df_exemple[['age']])

print("Données préprocessées :")
result = pd.concat([couleur_encoded, df_exemple[['satisfaction_encoded', 'age_scaled']]], axis=1)
print(result)   
```
