# Code de calcul des pourcentages dans différentes situations.

```python
def calculer_pourcentage(partie, total):
    """Calcule le pourcentage d'une partie par rapport à un total"""
    return (partie / total) * 100

def calculer_variation_pourcentage(ancienne, nouvelle):
    """Calcule la variation en pourcentage entre deux valeurs"""
    return ((nouvelle - ancienne) / ancienne) * 100

def points_de_pourcentage(pct1, pct2):
    """Calcule la différence en points de pourcentage"""
    return pct2 - pct1
 
# Exemples d'utilisation
print(f"Pourcentage de réussite : {calculer_pourcentage(20, 25):.1f}%")
print(f"Variation des ventes : {calculer_variation_pourcentage(50000, 65000):.1f}%")
print(f"Évolution en pp : {points_de_pourcentage(10, 12):.1f} pp")

# Attention aux variations successives
valeur_initiale = 100
valeur_apres_hausse = valeur_initiale * 1.10
valeur_finale = valeur_apres_hausse * 0.90
print(f"\nEffet de variations successives +10% puis -10% :")
print(f"Valeur finale : {valeur_finale:.2f} (soit {((valeur_finale/valeur_initiale)-1)*100:.1f}% de variation nette)")
```
