# Solutions des exercices sur le Scatter Plot

Ce notebook génère les scatter plots pour les 5 exercices en utilisant Python avec `matplotlib` et `pandas`, et décrit les relations observées.

```python
import matplotlib.pyplot as plt
import pandas as pd
```

## Exercice 1 : Santé et activité physique (Santé)

```python
# Données
data1 = pd.DataFrame({
    'Heures_activite': [2, 3, 5, 6, 4, 7, 1, 8, 3, 6],
    'Cholesterol_LDL': [150, 145, 130, 125, 140, 120, 160, 115, 148, 128]
})

# Scatter plot
plt.figure(figsize=(8, 6))
plt.scatter(data1['Heures_activite'], data1['Cholesterol_LDL'], color='blue')
plt.title('Heures d’activité vs Cholestérol LDL')
plt.xlabel('Heures d’activité par semaine')
plt.ylabel('Cholestérol LDL (mg/dL)')
plt.grid(True)
plt.savefig('ex1_scatter.png')
plt.show()

 
```

## Exercice 2 : Dépenses publicitaires et ventes (Économie)

```python
# Données
data2 = pd.DataFrame({
    'Depenses': [.5, 1.2, 2, .8, 1.8, .3, 2.5, 1.5,.9,2],
    'Ventes': [25, 58, 95, 35, 82, 15, 10, 70,42,88]
})

# Scatter plot
plt.figure(figsize=(8, 6))
plt.scatter(data2['Depenses'], data2['Ventes'], color='green')
plt.title('Dépenses publicitaires vs Ventes')
plt.xlabel('Dépenses publicitaires (milliers $)')
plt.ylabel('Ventes (milliers $)')
plt.grid(True)
plt.savefig('ex2_scatter.png')
plt.show()

 
```

## Exercice 3 : Propriétés immobilières

```python
# Données
data3 = pd.DataFrame({
    'surface': [45,65,85,55,75,35,95,105,40,80,90,60],
    'prix': [180,220,280,200,250,150,320,350,165,270,310,210]
})

# Scatter plot
plt.figure(figsize=(8, 6))
plt.scatter(data3['surface'], data3['prix'], color='red')
plt.title('Prix vs Surface')
plt.xlabel('Surface')
plt.ylabel('Prix')
plt.grid(True)
plt.savefig('ex3_scatter.png')
plt.show()

 
 
```

## Exercice 4 : Temps de chargement d'un site web

```python
# Données
data4 = pd.DataFrame({
    'chargement': [1.2,2.1,3.5,1.8,2.8,0.9,4.2,1.5,3.1,2.5,1.1,3.8,2.0,1.6,2.9],
    'rebond': [25,35,55,30,45,20,65,28,50,40,22,60,33,29,47]
})

# Scatter plot
plt.figure(figsize=(8, 6))
plt.scatter(data4['chargement'], data4['rebond'], color='purple')
plt.title('Taux de rebond vs Temps de chargement')
plt.xlabel('Temps de chargement')
plt.ylabel('Taux de rebond')
plt.grid(True)
plt.savefig('ex4_scatter.png')
plt.show()

 
```

## Exercice 5 : Evaluation de la qualité de prédiction avec un nuage à points

```python
# Données
data5 = pd.DataFrame({
    'salaire_reel': [35,45,55,40,60,30,70,50,65,38,75,42],
    'salaire_predit': [33,47,52,42,58,35,68,48,62,40,70,45 ]
})
import numpy as np
x_d = np.arange(25,80)
y_d = np.arange(25,80)

# Scatter plot
plt.figure(figsize=(8, 6))
plt.scatter(data5['salaire_reel'], data5['salaire_predit'], color='orange')
plt.plot(x_d, y_d, color='blue',
 linewidth=3)
plt.title('Salaire prédit vs Salaire réel')
plt.xlabel('Salaire réel')
plt.ylabel('Salaire prédit')
plt.grid(True)
plt.savefig('ex5_scatter.png')
plt.show()

 
```
