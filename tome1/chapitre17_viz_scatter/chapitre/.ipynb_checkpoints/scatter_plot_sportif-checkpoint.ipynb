```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Données des coureurs
data = {
    "Heures_entrainement": [2, 4, 3, 6, 5, 7, 8, 5, 9, 10],
    "Temps_100m": [15.2, 14.1, 14.5, 13.0, 13.6, 12.8, 12.5, 13.4, 12.1, 11.9]
}

df = pd.DataFrame(data)
df
```

```python
# Création du graphique en nuage de points
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="Heures_entrainement", y="Temps_100m", s=100, color="gray", edgecolor="black")
# plt.title("Relation entre les heures d'entraînement et le temps sur 100m")
plt.xlabel("Heures d'entraînement par semaine",fontsize=14)
plt.ylabel("Temps pour 100m (secondes)",fontsize=14)
plt.yticks(fontsize=12)
plt.xticks(fontsize=12)
plt.grid(True)
plt.tight_layout()
plt.savefig("scatter.png", dpi=300)
```
