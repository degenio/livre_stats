# Solutions des exercices sur le Scatter Plot

Ce notebook génère les scatter plots pour les 5 exercices en utilisant Python avec `matplotlib` et `pandas`, et décrit les relations observées.

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

## Exercice 1 : Santé et activité physique (Santé)

```python
import matplotlib.pyplot as plt

heures = [2, 3, 5, 6, 4, 7, 1, 8, 3, 6]
ldl = [150, 145, 130, 125, 140, 120, 160, 115, 148, 128]

r = np.corrcoef(heures, ldl)[0, 1]
print(f"r = {r:.3f}")
plt.figure(figsize=(6, 4))
plt.scatter(heures, ldl, color='black', s=50)
plt.title('Relation entre activité physique et cholestérol LDL')
plt.xlabel('Heures d’activité par semaine')
plt.ylabel('Cholestérol LDL (mg/dL)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.savefig('ex1_scatter.png')
plt.show()
```

## Exercice 2 : Dépenses publicitaires et ventes (Économie)

```python
import matplotlib.pyplot as plt

# Données
depenses = [0.5, 1.2, 2.0, 0.8, 1.8, 0.3, 2.51, 1.5, 0.9, 2.0]
conversions = [25, 58, 95, 35, 82, 15, 10, 70, 42, 88]
campagnes = ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J']

plt.figure(figsize=(8, 5))
plt.scatter(depenses, conversions, color='black', s=60)

# Annoter chaque point avec le nom de la campagne
for i, txt in enumerate(campagnes):
    plt.annotate(txt, (depenses[i], conversions[i]), textcoords="offset points", xytext=(5,5))

plt.title('Dépenses publicitaires vs Conversions')
plt.xlabel('Dépenses publicitaires (k$)')
plt.ylabel('Conversions')
plt.grid(True, linestyle='--', alpha=0.6)
plt.savefig('ex2_scatter.png')
plt.show()
```

## Exercice 3 : Propriétés immobilières

```python
import matplotlib.pyplot as plt

# Données
surface = [45, 65, 85, 55, 75, 35, 95, 105, 40, 80, 90, 60]
prix = [180, 220, 280, 200, 250, 150, 320, 350, 165, 270, 310, 210]
proprietes = list(range(1, 13))

plt.figure(figsize=(8, 5))
plt.scatter(surface, prix, color='black', s=60)

# Annoter chaque point avec le numéro de la propriété
for i, txt in enumerate(proprietes):
    plt.annotate(txt, (surface[i], prix[i]), textcoords="offset points", xytext=(5,5))

plt.title(' Prix de vente vs Surface habitable')
plt.xlabel('Surface (m²)')
plt.ylabel('Prix (k$)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.savefig('ex3_scatter.png')
plt.show() 
 
```

## Exercice 5 : Evaluation de la qualité de prédiction avec un nuage à points

```python
# Données
data5 = pd.DataFrame({
    'salaire_reel': [35,45,55,40,60,30,70,50,65,38,75,42],
    'salaire_predit': [33,47,52,42,58,35,68,48,62,40,70,45 ]
})
import numpy as np
x_d = np.arange(25,80)
y_d = np.arange(25,80)

# Scatter plot
plt.figure(figsize=(8, 6))
plt.scatter(data5['salaire_reel'], data5['salaire_predit'], color='gray')
plt.plot(x_d, y_d, color='black',
 linewidth=3)
plt.title('Salaire prédit vs Salaire réel')
plt.xlabel('Salaire réel')
plt.ylabel('Salaire prédit')
plt.grid(True)
plt.savefig('ex5_scatter.png')
plt.show()

 
```
