# Graphiques à barres

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Données exemple (ventes par filiale)
ventes = pd.DataFrame({
    'Filiale': ['Québec', 'Ontario', 'Colombie-Britannique', 'Alberta', 'Nouvelle-Écosse'],
    'Ventes': [320, 450, 380, 290, 210]
})

# 1. Barres simples- verticales (triées par valeur décroissante)
ventes_trie = ventes.sort_values('Ventes', ascending=False)
plt.figure(figsize=(8, 5))
sns.barplot(data=ventes_trie, x='Filiale', y='Ventes', palette='viridis',hue='Ventes', legend=False)
plt.title('Ventes de téléphones intelligents par filiale (avril 2025)')
plt.ylabel('Unités vendues')
plt.xlabel('Filiale')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
 
# 2. Barres horizontales (recommandé pour libellés longs)
plt.figure(figsize=(8, 5))
sns.barplot(data=ventes_trie, y='Filiale', x='Ventes', palette='viridis',hue='Ventes', legend=False)
plt.title('Ventes par filiale (orientation horizontale)')
plt.xlabel('Unités vendues')
plt.ylabel('Filiale')
plt.tight_layout()
plt.show()

# 3. Barres groupées (comparaison temporelle)
ventes_mensuelles = pd.DataFrame({
    'Filiale': ['Ontario', 'Ontario', 'Québec', 'Québec'],
    'Mois': ['Avril', 'Mai', 'Avril', 'Mai'],
    'Ventes': [450, 480, 320, 340]
})
plt.figure(figsize=(8, 5))
sns.barplot(data=ventes_mensuelles, x='Filiale', y='Ventes', hue='Mois', palette='Set2')
plt.title('Comparaison des ventes Avril vs Mai')
plt.ylabel('Unités vendues')
plt.tight_layout()
plt.show()

# 4. Barres empilées (composition par région)
composition = pd.DataFrame({
    'Produit': ['Pain', 'Banane', 'Ail', 'Carotte'],
    'Est': [40, 50, 30, 60],
    'Ouest': [30, 40, 45, 50]
})
composition_melted = composition.melt(id_vars='Produit', var_name='Région', value_name='Ventes')
plt.figure(figsize=(8, 5))
sns.barplot(data=composition_melted, x='Produit', y='Ventes', hue='Région', palette='Set2')
plt.title('Ventes par produit et région (empilées visuellement)')
plt.ylabel('Unités vendues')
plt.tight_layout()
plt.show()
```
