# Code de calcul du mode dans différentes situations.

```python
import numpy as np
from scipy import stats
import pandas as pd

# Données exemple (nombre de livres lus)
livres = np.array([5, 4, 3, 4, 6, 3, 2, 5, 4, 3, 2,
                   6, 5, 4, 3, 6, 5, 5, 4, 2, 3, 4, 5,
                   6, 6, 5, 3, 4, 5, 6])

# Mode avec scipy (gère le multimodal)
mode_result = stats.mode(livres, keepdims=False)
print(f"Mode : {mode_result.mode}, Fréquence : {mode_result.count}")

# Vérification manuelle avec pandas
s = pd.Series(livres)
frequences = s.value_counts()
print(f"\nTableau des fréquences :\n{frequences.sort_index()}")
print(f"\nMode(s) : {frequences[frequences == frequences.max()].index.tolist()}")

# Exemple bimodal
donnees_bimodales = np.array([1, 1, 1, 2, 2, 2, 3, 4, 5])
mode_bimodal = stats.mode(donnees_bimodales, keepdims=False)
print(f"\nExemple bimodal : mode = {mode_bimodal.mode} (fréquence {mode_bimodal.count})")
print("Note : scipy.stats.mode ne retourne qu'un seul mode ; utiliser value_counts() pour détecter le multimodal")
```
