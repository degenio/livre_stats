# Code de calcul de la moyenne dans différentes situations.

```python
import numpy as np
import pandas as pd

# Données exemple
notes = np.array([78, 82, 85, 88, 90, 92, 95, 98, 100, 100, 
                  100, 100, 92, 95, 100, 100, 100, 100, 68, 70, 
                  72, 75, 60, 62, 65])
 
# Moyenne simple
moyenne = np.mean(notes)
print(f"Moyenne simple : {moyenne:.2f}")

# Moyenne avec détection d'outliers (règle 3 écarts-types)
ecart_type = np.std(notes, ddof=1)
bornes = (moyenne - 3*ecart_type, moyenne + 3*ecart_type)
outliers = notes[(notes < bornes[0]) | (notes > bornes[1])]
print(f"Valeurs aberrantes potentielles : {outliers if len(outliers) > 0 else 'Aucune'}")

# Moyenne pondérée exemple
valeurs = np.array([85, 90, 75])  # Notes
poids = np.array([3, 2, 1])        # Coefficients
moyenne_ponderee = np.average(valeurs, weights=poids)
print(f"Moyenne pondérée : {moyenne_ponderee:.2f}")

# Comparaison moyenne vs médiane sur données asymétriques
donnees_asym = np.array([10, 20, 30, 1000])
print(f"\nDonnées asymétriques : {donnees_asym}")
print(f"Moyenne : {np.mean(donnees_asym):.2f}")
print(f"Médiane : {np.median(donnees_asym):.2f}")
```
