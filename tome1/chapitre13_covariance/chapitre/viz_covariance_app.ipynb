```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import pearsonr

# Simuler des données corrélées
np.random.seed(0)
n = 100

# Générer dépenses publicitaires
depenses = np.random.normal(50000, 10000, n)  # moyenne 50k$, écart-type 10k$

# Générer ventes corrélées avec un bruit
ventes = depenses * 0.6 + np.random.normal(0, 5000, n)  # coefficient ~0.6

# Calculer covariance empirique
cov_empirique = np.cov(depenses, ventes)[0,1]

# Calculer coefficient de corrélation et p-valeur
r, p_value = pearsonr(depenses, ventes)

print(f"Covariance empirique : {cov_empirique:.0f}")
print(f"Coefficient corrélation : {r:.3f} ")

# Visualisation
plt.figure(figsize=(6,3))
plt.scatter(depenses, ventes, alpha=0.7,color='gray')
# plt.title("Relation entre dépenses publicitaires et ventes")
plt.xlabel("Dépenses publicitaires ($)",fontsize=14)
plt.ylabel("Ventes ($)",fontsize=14)
plt.grid(True, linestyle='--', alpha=0.5)

plt.text(min(depenses), max(ventes)*0.95,
         f"Covariance = {cov_empirique:.0f}",
         fontsize=12, bbox=dict(facecolor='white', alpha=0.8))

plt.tight_layout()
plt.savefig("covariance_depenses_ventes.png", dpi=300)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import pearsonr

# Simuler les données
np.random.seed(42)
n = 100
age = np.random.normal(35, 12, n)  # moyenne d’âge 35 ans, σ = 12
temps_reseaux = 300 - 5 * age + np.random.normal(0, 30, n)  # relation négative + bruit

# Calcul de la covariance
covariance = np.cov(age, temps_reseaux)[0, 1]

# Corrélation de Pearson
r, p_value = pearsonr(age, temps_reseaux)

# Affichage
plt.figure(figsize=(6,3))
plt.scatter(age, temps_reseaux, color='gray', alpha=0.7)
# plt.title("Relation entre âge et temps passé sur les réseaux sociaux")
plt.xlabel("Âge (années)",fontsize=14)
plt.ylabel("Temps sur RS (minutes/jour)",fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)

# Annoter covariance et p-valeur
plt.text(40, max(temps_reseaux)*0.85,
         f"Covariance ≈ {covariance:.0f}",
         fontsize=11, bbox=dict(facecolor='white', alpha=0.8))

plt.tight_layout()
plt.savefig("covariance_age_reseaux.png", dpi=300)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import pearsonr

# Générer des données
np.random.seed(123)
n = 100
rendement_A = np.random.normal(0.05, 0.02, n)  # Action A : moyenne 5%, sigma 2%
rendement_B = rendement_A * 0.2 + np.random.normal(0.03, 0.02, n)  # faible dépendance + bruit

# Calcul de la covariance
cov = np.cov(rendement_A, rendement_B)[0, 1]

# Corrélation et p-value
r, p_value = pearsonr(rendement_A, rendement_B)

# Affichage
plt.figure(figsize=(6, 3))
plt.scatter(rendement_A, rendement_B, color='gray', alpha=0.7)
# plt.title("Relation entre les rendements de deux actions")
plt.xlabel("Rendement action A",fontsize=14)
plt.ylabel("Rendement action B",fontsize=14)
plt.grid(True, linestyle='--', alpha=0.5)

# Affichage des stats
plt.text(min(rendement_A), max(rendement_B)*0.85,
         f"Covariance ≈ {cov:.3f}",
         fontsize=11, bbox=dict(facecolor='white', alpha=0.8))

plt.tight_layout()
plt.savefig("covariance_actions.png", dpi=300)
plt.show()
```
