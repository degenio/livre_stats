# Code de calcul de la covariance dans différentes situations.

```python
import numpy as np
import pandas as pd

# Données exemple (revenu vs consommation)
donnees = pd.DataFrame({
    'revenu': [2000, 2500, 3000, 3500, 4000],
    'consommation': [1500, 1800, 2000, 2200, 2500]
})

# Covariance d'échantillon (ddof=1) avec pandas
cov_pandas = donnees['revenu'].cov(donnees['consommation'])
print(f"Covariance (pandas, ddof=1) : {cov_pandas:.2f}")

# Covariance avec numpy (retourne une matrice)
cov_numpy = np.cov(donnees['revenu'], donnees['consommation'], ddof=1)
print(f"Matrice de variance-covariance :\n{cov_numpy}")
print(f"Cov(X,Y) = {cov_numpy[0,1]:.2f}")
 
# Vérification manuelle
x_centre = donnees['revenu'] - donnees['revenu'].mean()
y_centre = donnees['consommation'] - donnees['consommation'].mean()
cov_manuel = (x_centre * y_centre).sum() / (len(donnees) - 1)
print(f"Covariance (calcul manuel) : {cov_manuel:.2f}")

# Corrélation pour comparaison
corr = donnees['revenu'].corr(donnees['consommation'])
print(f"Corrélation de Pearson : r = {corr:.3f}")

# Exemple de Cov=0 sans indépendance (Y = X^2)
print("\nContre-exemple : Cov=0 sans indépendance")
X = np.array([-2, -1, 0, 1, 2])
Y = X**2
print(f"Cov(X, X^2) = {np.cov(X, Y, ddof=1)[0,1]:.3f} (proche de 0)")
print(f"Corr(X, X^2) = {np.corrcoef(X, Y)[0,1]:.3f}")
```
