```python
import matplotlib.pyplot as plt

# Données statistiques manuelles
box_data = {
    'med': 15,         # Médiane (Q2)
    'q1': 12,          # Q1
    'q3': 18,          # Q3
    'whislo': 6,       # Min sans outliers
    'whishi': 22,      # Max sans outliers
    'fliers': [2, 25]  # Outliers
}

fig, ax = plt.subplots(figsize=(4, 4))

# Boxplot à partir des statistiques manuelles
ax.bxp([box_data], showfliers=True, patch_artist=True,
       boxprops=dict(facecolor='gray'),
       flierprops=dict(marker='o', markerfacecolor='gray', markersize=10),
       medianprops=dict(color='black', linewidth=2))

# Mise en forme
ax.set_ylabel("Années d'éducation",fontsize=14)
ax.set_xticks([1])
ax.set_xticklabels(["Population"],fontsize=14)
# ax.set_title("Distribution des années d'éducation")

# Annotations
ax.annotate("Q1: 12", xy=(1, 12), xytext=(1.15, 12), textcoords='data', color='black',fontsize=14)
ax.annotate("Q2: 15", xy=(1, 15), xytext=(1.15, 15), textcoords='data', color='black',fontsize=14)
ax.annotate("Q3: 18", xy=(1, 18), xytext=(1.15, 18), textcoords='data', color='black',fontsize=14)

plt.tight_layout()
plt.savefig("boxplot_education_quartiles.png", dpi=300)
plt.show()
```

```python
import matplotlib.pyplot as plt

# Statistiques manuelles pour le boxplot
box_data = {
    'med': 7,         # Médiane
    'q1': 5,          # Q1
    'q3': 9,          # Q3
    'whislo': 3,      # Min sans outliers
    'whishi': 12,     # Max sans outliers
    'fliers': [2, 16] # Valeurs atypiques
}

fig, ax = plt.subplots(figsize=(4, 4))

# Création du boxplot
ax.bxp([box_data], showfliers=True, patch_artist=True,
       boxprops=dict(facecolor='gray'),
       flierprops=dict(marker='o', markerfacecolor='gray', markersize=8),
       medianprops=dict(color='black', linewidth=2))

# Mise en forme
ax.set_ylabel("Temps de récupération (jours)",fontsize=14)
ax.set_xticks([1])
ax.set_xticklabels(["Post-op"],fontsize=14)
# ax.set_title("Diagramme en boîte des temps de récupération")

# Annotations pour les quartiles
ax.annotate("Q1: 5 j", xy=(1, 5), xytext=(1.15, 5), textcoords='data', fontsize=14)
ax.annotate("Q2: 7 j", xy=(1, 7), xytext=(1.15, 7), textcoords='data', fontsize=14)
ax.annotate("Q3: 9 j", xy=(1, 9), xytext=(1.15, 9), textcoords='data', fontsize=14)

plt.tight_layout()
plt.savefig("boxplot_temps_recuperation.png", dpi=300)
plt.show()
```

```python
import matplotlib.pyplot as plt

# Statistiques manuelles simulées
box_data = {
    'med': 0.3,           # Médiane
    'q1': -1.2,           # Q1
    'q3': 1.4,            # Q3
    'whislo': -3.0,       # Min sans outliers
    'whishi': 3.0,        # Max sans outliers
    'fliers': [-6.0, 5.5] # Outliers
}

fig, ax = plt.subplots(figsize=(4, 4))

# Création du boxplot
ax.bxp([box_data], showfliers=True, patch_artist=True,
       boxprops=dict(facecolor='gray'),
       flierprops=dict(marker='o', markerfacecolor='gray', markersize=10),
       medianprops=dict(color='black', linewidth=2))

# Mise en forme
ax.set_ylabel("Rendement journalier (%)",fontsize=14)
ax.set_xticks([1])
ax.set_xticklabels(["Bourse"],fontsize=14)
# ax.set_title("Quartiles des rendements boursiers")

# Annotations
ax.annotate("Q1: -1.2%", xy=(1, -1.2), xytext=(1.15, -1.2), textcoords='data',fontsize=14)
ax.annotate("Q2:  0.3%", xy=(1, 0.3), xytext=(1.15, 0.3), textcoords='data',fontsize=14)
ax.annotate("Q3:  1.4%", xy=(1, 1.4), xytext=(1.15, 1.4), textcoords='data',fontsize=14)

plt.tight_layout()
plt.savefig("boxplot_rendements_boursiers.png", dpi=300)
plt.show()
```
