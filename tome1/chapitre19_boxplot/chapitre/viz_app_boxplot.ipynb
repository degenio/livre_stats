```python
import matplotlib.pyplot as plt
import numpy as np

# Données modifiées pour forcer des outliers visibles
rendements = np.concatenate([
    np.random.normal(loc=2.0, scale=0.5, size=30),  # données principales
    [4.0, 4.5, -1.5]  # outliers (volontaires)
])

fig, ax = plt.subplots(figsize=(4, 4))

# Créer le boxplot
box = ax.boxplot(
    rendements,
    vert=True,
    patch_artist=True,
    boxprops=dict(facecolor='gray'),
    medianprops=dict(color='black', linewidth=2),
    flierprops=dict(marker='o', markerfacecolor='black', markersize=10, linestyle='none')
)

# Extraire et étiqueter les outliers
outliers_x = box["fliers"][0].get_xdata()
outliers_y = box["fliers"][0].get_ydata()

for x, y in zip(outliers_x, outliers_y):
    ax.annotate(f"{y:.1f}",
                xy=(x, y),
                xytext=(8, 0),
                textcoords="offset points",
                ha='left',
                va='center',
                fontsize=10,
                color='black')

# Mise en page
ax.set_ylabel("Rendement quotidien (%)",fontsize=14)
ax.set_xticks([1])
ax.set_xticklabels(["Rendements boursiers"],fontsize=14)
# ax.set_title("Diagramme en boîte vertical avec outliers")
plt.yticks(fontsize=12)
plt.xticks(fontsize=12)
plt.tight_layout()
plt.savefig("rendements_boxplot_vertical_outliers.png", dpi=300)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Données simulées : temps de récupération en jours
temps_recup = np.concatenate([
    np.random.normal(loc=7, scale=1.5, size=30),  # récupération typique
    [15, 17]  # cas prolongés anormaux
])

fig, ax = plt.subplots(figsize=(4, 4))  # Format vertical

# Création du boxplot vertical
box = ax.boxplot(
    temps_recup,
    vert=True,
    patch_artist=True,
    boxprops=dict(facecolor='gray'),
    medianprops=dict(color='black', linewidth=2),
    flierprops=dict(marker='o', markerfacecolor='black', markersize=10, linestyle='none')
)

# Étiquetage des outliers
outliers_x = box["fliers"][0].get_xdata()
outliers_y = box["fliers"][0].get_ydata()

for x, y in zip(outliers_x, outliers_y):
    ax.annotate(f"{y:.1f} j",
                xy=(x, y),
                xytext=(8, 0),
                textcoords="offset points",
                ha='left',
                va='center',
                fontsize=10,
                color='black')

# Mise en forme du graphique
ax.set_ylabel("Temps de récupération (jours)",fontsize=14)
ax.set_xticks([1])
ax.set_xticklabels(["Post-opératoire"],fontsize=14)
# ax.set_title("Diagramme en boîte des temps de récupération")
plt.yticks(fontsize=12)
plt.xticks(fontsize=12)
plt.tight_layout()
plt.savefig("temps_recuperation_boxplot.png", dpi=300)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Données simulées : niveaux de pollution PM2.5 en µg/m³ pour plusieurs villes
np.random.seed(42)
villes = ["Montréal", "Paris", "Delhi", "Stockholm"]

pollution_data = [
    np.random.normal(loc=12, scale=3, size=30),   # Montréal
    np.random.normal(loc=18, scale=5, size=30),   # Paris
    np.concatenate([np.random.normal(loc=60, scale=10, size=28), [110, 120]]),  # Delhi avec outliers
    np.random.normal(loc=8, scale=2, size=30)     # Stockholm
]

fig, ax = plt.subplots(figsize=(6, 5))

# Création du boxplot
box = ax.boxplot(
    pollution_data,
    patch_artist=True,
    labels=villes,
    flierprops=dict(marker='o', markerfacecolor='black', markersize=10, linestyle='none')
)

# Couleur des boîtes
colors = ['lightgray', 'gray', 'gray', 'lightgray']
for patch, color in zip(box['boxes'], colors):
    patch.set_facecolor(color)

# Annoter les outliers de Delhi (ville 3)
outliers_x = box["fliers"][2].get_xdata()
outliers_y = box["fliers"][2].get_ydata()
for x, y in zip(outliers_x, outliers_y):
    ax.annotate(f"{y:.0f}",
                xy=(x, y),
                xytext=(8, 0),
                textcoords="offset points",
                ha='left',
                va='center',
                fontsize=10,
                color='black')

# Mise en forme
ax.set_ylabel("PM2.5 (µg/m³)",fontsize=14)
# ax.set_title("Niveaux de pollution PM2.5 par ville")
plt.yticks(fontsize=12)
plt.xticks(fontsize=12)
plt.tight_layout()
plt.savefig("pollution_boxplot_villes.png", dpi=300)
plt.show()
```
