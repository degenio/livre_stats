# Nuages de points

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

# Données exemple (heures d'entraînement vs temps 100m)
donnees = pd.DataFrame({
    'heures': [2, 4, 3, 6, 5, 7, 8, 5, 9, 10],
    'temps_100m': [15.2, 14.1, 14.5, 13.0, 13.6, 12.8, 12.5, 13.4, 12.1, 11.9]
})

# 1. Nuage de points simple avec régression linéaire
plt.figure(figsize=(8, 6))
sns.regplot(data=donnees, x='heures', y='temps_100m', 
            scatter_kws={'alpha': 0.7}, line_kws={'color': 'red'})
 
plt.title('Temps sur 100m en fonction des heures d\'entraînement')
plt.xlabel('Heures d\'entraînement par semaine')
plt.ylabel('Temps sur 100m (secondes)')
plt.tight_layout()
plt.show()

# 2. Nuage avec encodage par catégorie (bubble plot)
donnees_cat = pd.DataFrame({
    'depenses': [10, 15, 20, 18, 40, 50, 55, 60, 90, 95, 100, 110],
    'ventes': [80, 100, 120, 110, 250, 270, 300, 280, 480, 500, 520, 550],
    'categorie': ['A']*4 + ['B']*4 + ['C']*4,
    'roi': [8, 6.7, 6, 6.1, 6.25, 5.4, 5.45, 4.67, 5.33, 5.26, 5.2, 5]  # variable pour taille
})

plt.figure(figsize=(8, 6))
sns.scatterplot(data=donnees_cat, x='depenses', y='ventes', 
                hue='categorie', size='roi', sizes=(20, 200),
                palette='Set2', alpha=0.7)
plt.title('Ventes vs Dépenses publicitaires par catégorie')
plt.xlabel('Dépenses publicitaires ($)')
plt.ylabel('Ventes ($)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()

# 3. Gestion des grands volumes : hexbin plot
np.random.seed(42)
n = 10000
x_grand = np.random.normal(0, 1, n)
y_grand = 2*x_grand + np.random.normal(0, 0.5, n)

plt.figure(figsize=(8, 6))
plt.hexbin(x_grand, y_grand, gridsize=30, cmap='viridis', mincnt=1)
plt.colorbar(label='Nombre de points')
plt.xlabel('Variable X')
plt.ylabel('Variable Y')
plt.title('Hexbin plot pour 10 000 points')
plt.tight_layout()
plt.show()

# 4. Matrice de nuages de points (pairplot) pour EDA multivariée
iris = sns.load_dataset('iris')
sns.pairplot(iris, hue='species', diag_kind='hist', plot_kws={'alpha': 0.6})
plt.suptitle('Matrice de nuages de points - Dataset Iris', y=1.02)
plt.show()
```
