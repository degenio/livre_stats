# Histogrammes

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

# Données exemple (taux d'alcalinité)
alcalinite = np.array([87, 91, 85, 78, 94, 96, 83, 82, 89, 88,
                       92, 97, 95, 86, 90, 93, 84, 80, 81, 88,
                       90, 91, 92, 85, 84, 89, 90, 93, 94, 95])

# 1. Histogramme en fréquence (bins automatiques)
plt.figure(figsize=(8, 5))
plt.hist(alcalinite, bins='auto', edgecolor='black', alpha=0.7)
plt.title('Histogramme en fréquence - Alcalinité de l\'eau')
plt.xlabel('Alcalinité (mg/L)')
plt.ylabel('Effectif')
plt.grid(axis='y', alpha=0.5)
plt.tight_layout()
plt.show()
 
# 2. Histogramme en densité avec courbe KDE
plt.figure(figsize=(8, 5))
sns.histplot(alcalinite, bins=6, stat='density', kde=True, 
             edgecolor='black', alpha=0.7, color='skyblue')
plt.title('Histogramme en densité avec estimation KDE')
plt.xlabel('Alcalinité (mg/L)')
plt.ylabel('Densité')
plt.tight_layout()
plt.show()

# 3. Comparaison des règles de choix de bins
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
rules = [('sturges', 'Sturges'), ('sqrt', 'Racine carrée'), 
         ('scott', 'Scott'), ('fd', 'Freedman-Diaconis')]

for ax, (rule, name) in zip(axes.flat, rules):
    ax.hist(alcalinite, bins=rule, edgecolor='black', alpha=0.7)
    ax.set_title(f'Règle de {name}')
    ax.set_xlabel('Alcalinité (mg/L)')
    ax.set_ylabel('Effectif')
    ax.grid(axis='y', alpha=0.5)

plt.tight_layout()
plt.show()

# 4. Impact de l'origine des classes (décalage)
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
for ax, offset in zip(axes, [0, 2]):
    bins = np.arange(78 + offset, 98 + offset, 4)
    ax.hist(alcalinite, bins=bins, edgecolor='black', alpha=0.7)
    ax.set_title(f'Origine des classes : {78 + offset} mg/L')
    ax.set_xlabel('Alcalinité (mg/L)')
    ax.set_ylabel('Effectif')
    ax.grid(axis='y', alpha=0.5)
plt.tight_layout()
plt.show()
```
