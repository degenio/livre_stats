# Statistiques 101 : Solutions des exercices sur le boxplot
Ce notebook génère les boxplots pour les exercices et répond aux questions.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("seaborn-v0_8-pastel")
 
```

## Exercice 1 : Salaires par département

```python
data1 = {
    'Département': ['Marketing']*7 + ['Développement']*7 + ['Support client']*7,
    'Salaire': [45, 50, 55, 60, 65, 70, 95, 60, 65, 70, 75, 80, 85, 90, 35, 40, 45, 50, 55, 60, 65]
}
df1 = pd.DataFrame(data1)
plt.figure(figsize=(8,6))
sns.boxplot(x='Département', y='Salaire', data=df1,color='lightgray')
plt.title('Salaires par département')
plt.yticks(fontsize=12)
plt.xticks(fontsize=12)
plt.ylabel("Salaire",fontsize=14)
plt.xlabel("Départements",fontsize=14)
plt.savefig('ex1_boxplot.png')
plt.show()
```
