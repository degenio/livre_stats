# Code de l'ecart  type

```python
import numpy as np
import pandas as pd

# Données exemple (notes d'étudiants)
notes = np.array([28, 32, 35, 40, 42, 45, 48, 50, 52, 55, 
                  57, 60, 62, 65, 68, 70, 72, 75, 78, 82, 
                  85, 88, 90, 94, 100])

# Écart-type d'échantillon (ddof=1 = n-1)
std_echantillon = np.std(notes, ddof=1)
print(f"Écart-type (échantillon) : {std_echantillon:.2f}")

# Écart-type de population (ddof=0 = N)
std_population = np.std(notes, ddof=0)
print(f"Écart-type (population) : {std_population:.2f}")
 
# Coefficient de variation
moyenne = np.mean(notes)
cv = (std_echantillon / moyenne) * 100
print(f"Coefficient de variation : {cv:.1f}%")

# Vérification règle empirique (si distribution approximativement normale)
print(f"\nRègle empirique (si normalité) :")
print(f"  68% dans [{moyenne - std_echantillon:.2f}, {moyenne + std_echantillon:.2f}]")
print(f"  95% dans [{moyenne - 2*std_echantillon:.2f}, {moyenne + 2*std_echantillon:.2f}]")

# Avec pandas
s = pd.Series(notes)
print(f"\nAvec pandas : std = {s.std():.2f} (ddof=1 par défaut)")
```
