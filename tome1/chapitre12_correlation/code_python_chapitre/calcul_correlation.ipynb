# Code de calcul des coefficients de corrélation dans différentes situations.

```python
import numpy as np
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt

# Données exemple (temps d'étude vs notes)
donnees = pd.DataFrame({
    'temps_etude': [2, 3, 4, 5, 6],
    'notes': [60, 70, 75, 85, 90]
})

# Corrélation de Pearson
r, p_valor = stats.pearsonr(donnees['temps_etude'], donnees['notes'])
print(f"Pearson : r = {r:.3f}, p-value = {p_valor:.4f}")

# Corrélation de Spearman
rho, p_valor_s = stats.spearmanr(donnees['temps_etude'], donnees['notes'])
print(f"Spearman : rho = {rho:.3f}, p-value = {p_valor_s:.4f}")

# Coefficient de détermination
r2 = r**2
print(f"Coefficient de détermination R² = {r2:.3f} ({r2*100:.1f}% de variance expliquée)")

# Visualisation
plt.scatter(donnees['temps_etude'], donnees['notes'], alpha=0.7)
plt.xlabel('Temps d\'étude (heures)')
plt.ylabel('Notes')
plt.title(f'Corrélation : r = {r:.2f}')
plt.grid(True, alpha=0.3)
plt.show()
```
