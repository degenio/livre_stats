# Exercice 12.1


Un chercheur souhaite savoir s’il existe une corrélation entre le nombre d’heures d’étude par semaine et les
résultats aux examens finaux chez des étudiants universitaires. Il collecte les données de 10 étudiants.

Nous utiliserons la **corrélation de Pearson** pour mesurer la force et la direction de cette relation.


```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import pearsonr, spearmanr
```

```python

# Données  
data = {
    'Heures d’étude': [5,10,3,8,12,4,6,11,7,9],
    'Note finale': [65,78,55,74,88,58,68,85,70,76]
}
df = pd.DataFrame(data)
df
```

```python
# Nuage de points
ax=sns.scatterplot(data=df, x='Heures d’étude', y='Note finale',color='black')
ax.set_xlabel("Heures d’étude", fontsize=16)
ax.set_ylabel("Note finale", fontsize=16)
ax.tick_params(axis='both', which='major', labelsize=14)
plt.title("Relation entre Heures d’étude et Note finale")
plt.grid(True)
# plt.show()
plt.tight_layout()
plt.savefig('exercice121.png')
```

```python

# Calcul de la corrélation de Pearson
corr, p_value = pearsonr(df['Heures d’étude'], df['Note finale'])
print(f"Coefficient de corrélation de Pearson : {corr:.3f}")
print(f"P-value : {p_value:.4f}")
```

# Exercice 12.5

Un psychologue industriel étudie la relation entre le stress au travail (échelle 1-10) et la satisfaction professionnelle (échelle 1-100) chez 15 employés.

```python
# Données  
data = {
    'Stress (1-10)': [8,3,6,9,4,7,2,5,8,3,6,9,4,7,5],
    "Satisfaction (1-100)": [35,85,55,25,78,45,90,65,30,82,58, 28,75,48,68]
}
df = pd.DataFrame(data)
df
```

```python
# Nuage de points
ax = sns.scatterplot(data=df, x='Stress (1-10)', y="Satisfaction (1-100)",  color='black')
ax.set_xlabel("Stress (1-10)", fontsize=16)
ax.set_ylabel("Satisfaction (1-100)", fontsize=16)
ax.tick_params(axis='both', which='major', labelsize=14)
plt.title('Relation entre le niveau de stress et la satisfaction')
plt.grid(True)
# plt.show()
plt.tight_layout()
plt.savefig('exercice125.png')
# Calcul de la corrélation de Pearson
corr, p_value = pearsonr(df['Stress (1-10)'], df["Satisfaction (1-100)"])
print(f"Coefficient de corrélation de Pearson : {corr:.3f}")
print(f"P-value : {p_value:.4f}")
spearman_corr = df['Stress (1-10)'].corr(df['Satisfaction (1-100)'], method='spearman')
print(f"Coefficient de corrélation de Spearman : {spearman_corr:.3f}")
```
