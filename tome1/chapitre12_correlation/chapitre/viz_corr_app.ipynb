```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import pearsonr

# Données simulées : prix entre 800$ et 1000$
np.random.seed(0)
n = 50
prix = np.linspace(800, 1000, n)
# Demande décroît avec le prix + un peu de bruit
demande = 1500 - 1.2 * prix + np.random.normal(0, 30, n)

# Corrélation de Pearson
r, p_value = pearsonr(prix, demande)

# Affichage
plt.figure(figsize=(6,3))
plt.scatter(prix, demande, color='gray', alpha=0.7)
plt.plot(prix, np.poly1d(np.polyfit(prix, demande, 1))(prix), color='black', linestyle='--', label='Tendance')

# plt.title("Corrélation entre prix et demande")
plt.xlabel("Prix du produit ($)",fontsize=14)
plt.ylabel("Demande (unités)",fontsize=14)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()

# Annotations
plt.text(min(prix), max(demande)*0.5,
         f"$\\rho$ = {r:.2f}",
         fontsize=12,
         bbox=dict(facecolor='white', edgecolor='gray', alpha=0.8))

plt.tight_layout()
plt.savefig("correlation_prix_demande.png", dpi=300)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import pearsonr

# Simulation de données
np.random.seed(42)
n = 50
temps_etude = np.random.normal(5, 2, n)  # heures d’étude par jour
temps_etude = np.clip(temps_etude, 1, 10)  # pour rester dans un intervalle réaliste

# Génération des notes : note = a * temps + bruit
notes = 10 * temps_etude + np.random.normal(0, 5, n)

# Corrélation de Pearson
r, p_value = pearsonr(temps_etude, notes)

# Tracé
plt.figure(figsize=(6,3))
plt.scatter(temps_etude, notes, color='gray', alpha=0.7, label='Données')
plt.plot(np.unique(temps_etude),
         np.poly1d(np.polyfit(temps_etude, notes, 1))(np.unique(temps_etude)),
         color='black', linestyle='--', label='Régression')

# plt.title("Corrélation entre temps d’étude et notes")
plt.xlabel("Temps d’étude (heures/jour)",fontsize=14)
plt.ylabel("Note à l'examen",fontsize=14)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()

# Annotations statistiques
plt.text( 1,max(notes)*0.75, 
         f"$r = {r:.2f}$",
         fontsize=12,  
         bbox=dict(facecolor='white', edgecolor='gray', alpha=0.8))

plt.tight_layout()
plt.savefig("correlation_etude_notes.png", dpi=300)
plt.show()
```
