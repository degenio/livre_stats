# Code Python suivant permet de tracer les visualisations de base.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Données exemple
np.random.seed(42)
donnees = pd.DataFrame({'categorie': ['A']*30 + ['B']*30 + ['C']*30,
    'valeur': np.concatenate([np.random.normal(10, 2, 30), np.random.normal(12, 3, 30), np.random.normal(8, 1.5, 30)]),'temps': np.tile(np.arange(30), 3)})

# 1. Histogramme
plt.figure(figsize=(5, 4))
sns.histplot(donnees['valeur'], bins=15, kde=True)
plt.title('Distribution d\'une variable quantitative')
plt.xlabel('Valeur')
plt.ylabel('Fréquence')
plt.tight_layout()
plt.show()

# 2. Boxplot comparatif
plt.figure(figsize=(5, 4))
sns.boxplot(x='categorie', y='valeur', data=donnees)
plt.title('Comparaison de distributions par catégorie')
plt.tight_layout()
plt.show()

# 3. Nuage de points avec régression
plt.figure(figsize=(5, 4))
sns.regplot(x='temps', y='valeur', 
            data=donnees[donnees['categorie']=='A'], 
            scatter_kws={'alpha':0.6})
plt.title('Relation temporelle avec tendance linéaire')
plt.tight_layout()
plt.show()
 
# 4. Diagramme en barres
counts = donnees['categorie'].value_counts()
plt.figure(figsize=(5, 4))
counts.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Effectifs par catégorie')
plt.xlabel('Catégorie')
plt.ylabel('Nombre d\'observations')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
