# Graphes à barres  

Ce notebook génère le graphes à barres de la Figure 15.6 – Ventes de téléphones intelligents par filiale.

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# Configuration générale pour les graphes
province = ['Ontario', 'Colombie-britannique', 'Québec', 'Alberta', 'Nouvelle-Écosse']
ventes = [450, 380, 320, 290, 210]

plt.figure(figsize=(8, 6))
 
sns.barplot(
    x=province, y=ventes, color='gray' )
 
plt.xlabel('Filiale', fontsize=14)
plt.ylabel('Nombre de téléphones vendus', fontsize=14)
# plt.title('Ventes de téléphones intelligents par filiale.')
# plt.grid(True)
plt.yticks(fontsize=12)
plt.xticks(fontsize=12)
plt.tight_layout()
plt.savefig('graphique_barres_ventes.png')
 

```
