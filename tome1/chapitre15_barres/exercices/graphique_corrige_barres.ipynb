# Graphes à barres  

Ce notebook génère les graphes à barres pour les 5 exercices sur les graphes à barres. Chaque graphe est créé avec Matplotlib et sauvegardé comme une image PNG.

```python
import matplotlib.pyplot as plt
import numpy as np

 

# Exercice 1 : Répartition des étudiants par filière
filieres = ['Informatique','Biologie', 'Mathématiques',  'Physique',  'Chimie']
etudiants = [180,150, 120, 90, 60]

plt.figure(figsize=(8, 6))
# plt.grid(which='minor', color='#EEEEEE', linestyle='-', linewidth=0.5,zorder=0)
plt.minorticks_on()
plt.bar(filieres, etudiants, color='gray')
plt.xlabel('Filière',fontsize=14)
plt.ylabel('Nombre d’étudiants',fontsize=14)
plt.title('Répartition des étudiants par filière')
# plt.grid(True)
plt.yticks(fontsize=12)
plt.xticks(fontsize=12)
 
plt.legend()
plt.tight_layout()
plt.savefig('exercice1_barres.png')
 
```

```python
# Exercice 2 : Consommation d'énergie par source
# sources = ['Hydraulique','Éolienne', 'Nucléaire','Solaire',  'Charbon']
# pourcentages = [30,25, 20,15,  10]
# color= ['W', 'X', 'Y', 'Z','B']
import pandas as pd
import seaborn as sns

data = {
    'sources': ['Hydraulique','Éolienne', 'Nucléaire','Solaire',  'Charbon'],
    'pourcentages': [30,25, 20,15,  10] 
      
}
df = pd.DataFrame(data)
sns.light_palette('blue')
plt.figure(figsize=(8, 6))
sns.barplot(x='sources', y='pourcentages',     saturation=0.3,  palette=['gray', 'gray', 'gray','gray', 'gray'],data=df)
# plt.bar(sources, pourcentages, color='lightgreen')
plt.xlabel('Source d’énergie',fontsize=14)
plt.ylabel('Pourcentage (%)',fontsize=14)
plt.yticks(fontsize=12)
plt.xticks(fontsize=12)
plt.title('Consommation d’énergie par source')
# plt.grid(True) 
plt.tight_layout()
plt.savefig('exercice2_barres.png')
 
```

```python
# Exercice 3 : Préférences alimentaires dans un hôpital
repas = ['Cacher', 'Halal','Sans gluten','Végétarien','Standard'     ]
patients = [10,20,30,50,90]

plt.figure(figsize=(8, 6))
plt.barh(repas, patients, color='gray')
plt.ylabel('Type de repas',fontsize=14)
plt.xlabel('Nombre de patients',fontsize=14)
plt.yticks(fontsize=12)
plt.xticks(fontsize=12)
plt.title('Préférences alimentaires des patients')

plt.tight_layout()
plt.savefig('exercice3_barres.png')
 
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'joueurs': ['Simon', 'Albert', 'Abdel', 'Jean', 'David'],
    'tentative': [25, 20, 18, 15, 10],
    'reussi': [15, 12, 10, 8, 5]
}

df = pd.DataFrame(data)

# Graphique à barres empilées en niveaux de gris
ax = df.plot(
    x='joueurs',
    kind='bar',
    stacked=True,
    color=['#4D4D4D', '#B3B3B3'],  # gris foncé et gris clair
    figsize=(8, 6)
)

ax.set_xlabel('Joueur', fontsize=14)
ax.set_ylabel('Nombre de tirs', fontsize=14)
ax.set_title('Performance des joueurs au basketball', fontsize=16)

ax.legend(['Tentatives', 'Réussis'])
plt.title('Performance des joueurs au basket-ball')
plt.tight_layout()
plt.savefig('exercice4_barres.png', dpi=300)
plt.show()
```

```python
# Exercice 5 : Déséquilibre de classes
classes = ['Normal', 'Fraude']
effectifs = [850,150]
colors = ['Black', 'gray']  # Normal=850 → blue, Fraude=150 → red
plt.figure(figsize=(8, 6))
plt.bar(classes, effectifs, color=colors)
plt.axhline(y=500, linestyle='--')
plt.xlabel('Classe',fontsize=14)
plt.ylabel('Effectifs',fontsize=14)
plt.title('Déséquilibre de classes')
 
plt.tight_layout()
plt.savefig('exercice5_barres.png')
 
```
