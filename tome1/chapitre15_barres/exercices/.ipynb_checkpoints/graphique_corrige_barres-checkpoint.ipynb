# Graphes à barres  

Ce notebook génère les graphes à barres pour les 5 exercices sur les graphes à barres. Chaque graphe est créé avec Matplotlib et sauvegardé comme une image PNG.

```python
import matplotlib.pyplot as plt
import numpy as np

 

# Exercice 1 : Répartition des étudiants par filière
filieres = ['Informatique','Biologie', 'Mathématiques',  'Physique',  'Chimie']
etudiants = [180,150, 120, 90, 60]

plt.figure(figsize=(8, 6))
# plt.grid(which='minor', color='#EEEEEE', linestyle='-', linewidth=0.5,zorder=0)
plt.minorticks_on()
plt.bar(filieres, etudiants, color='skyblue')
plt.xlabel('Filière')
plt.ylabel('Nombre d’étudiants')
plt.title('Répartition des étudiants par filière')
# plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig('exercice1_barres.png')
 
```

```python
# Exercice 2 : Consommation d'énergie par source
# sources = ['Hydraulique','Éolienne', 'Nucléaire','Solaire',  'Charbon']
# pourcentages = [30,25, 20,15,  10]
# color= ['W', 'X', 'Y', 'Z','B']
import pandas as pd
import seaborn as sns

data = {
    'sources': ['Hydraulique','Éolienne', 'Nucléaire','Solaire',  'Charbon'],
    'pourcentages': [30,25, 20,15,  10] 
      
}
df = pd.DataFrame(data)
sns.light_palette('blue')
plt.figure(figsize=(8, 6))
sns.barplot(x='sources', y='pourcentages',     saturation=0.3,  palette=['blue', 'green', 'red','yellow', 'black'],data=df)
# plt.bar(sources, pourcentages, color='lightgreen')
plt.xlabel('Source d’énergie')
plt.ylabel('Pourcentage (%)')
plt.title('Consommation d’énergie par source')
# plt.grid(True) 
plt.tight_layout()
plt.savefig('exercice2_barres.png')
 
```

```python
# Exercice 3 : Préférences alimentaires dans un hôpital
repas = ['Cacher', 'Halal','Sans gluten','Végétarien','Standard'     ]
patients = [10,20,30,50,90]

plt.figure(figsize=(8, 6))
plt.barh(repas, patients, color='salmon')
plt.ylabel('Type de repas')
plt.xlabel('Nombre de patients')
plt.title('Préférences alimentaires des patients')

plt.tight_layout()
plt.savefig('exercice3_barres.png')
 
```

```python
# Exercice 5 : Déséquilibre de classes
classes = ['Normal', 'Fraude']
effectifs = [850,150]
colors = ['Cyan', 'salmon']  # Normal=850 → blue, Fraude=150 → red
plt.figure(figsize=(8, 6))
plt.bar(classes, effectifs, color=colors)
plt.axhline(y=500, linestyle='--')
plt.xlabel('Classe')
plt.ylabel('Effectifs')
plt.title('Déséquilibre de classes')
 
plt.tight_layout()
plt.savefig('exercice5_barres.png')
 
```
