# Histogramme du taux d'alcalinité de l'eau
Ce notebook affiche un histogramme basé sur les mesures du taux d'alcalinité.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

alcalinite = [
   87, 91, 85, 78, 94, 96, 83, 82, 89, 88,
92, 97, 95, 86, 90, 93, 84, 80, 81, 88,
90, 91, 92, 85, 84, 89, 90, 93, 94, 95
]

df = pd.DataFrame({'Taux d\'alcalinité (mg/L)': alcalinite})

plt.figure(figsize=(8, 6))
sns.histplot(df, x='Taux d\'alcalinité (mg/L)', bins=5, kde=False, alpha=0.6,color='gray', edgecolor='black')
# plt.title("Histogramme du taux d'alcalinité de l'eau")
plt.xlabel("Taux d'alcalinité (mg/L)",fontsize=14)
plt.grid(which='minor', color='#EEEEEE', linestyle='-', linewidth=0.5,zorder=0)
# Make the minor ticks and gridlines show.
# plt.minorticks_on()
plt.ylabel("Effectif",fontsize=14)
# plt.grid(True)
plt.yticks(fontsize=12)
plt.xticks(fontsize=12)
plt.tight_layout()
plt.savefig("histo.png", dpi=300)
```
