# Solutions des exercices sur l'Histogramme

Ce notebook génère les histogrammes pour les 5 exercices en utilisant Python avec `matplotlib` et `pandas`, et décrit les distributions observées.

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

## Exercice 1 : Durée de sommeil (Santé)

```python
# Données
data1 = [6.5, 7.0, 8.0, 6.8, 7.5, 6.2, 8.5, 7.2, 6.0, 7.8, 8.2, 6.7, 7.4, 6.9, 8.0, 7.1, 6.3, 8.3, 7.6, 6.4]

# Histogramme
plt.figure(figsize=(8, 6))
plt.hist(data1, bins=5, edgecolor='black')
plt.title('Histogramme de la durée de sommeil')
plt.xlabel('Durée de sommeil (heures)')
plt.ylabel('Fréquence')
plt.grid(True, alpha=0.3)
plt.savefig('ex1_histogramme.png')
plt.show()

# Interprétation
print('Forme : Assez symétrique, légèrement en cloche.')
print('Étendue : De 6.0 à 8.5 heures.')
print('Concentration : Pic autour de 6.5–7.5 heures.')
```

## Exercice 2 : Dépenses mensuelles (Économie)

```python
# Données
data2 = [120, 150, 200, 180, 130, 170, 210, 140, 160, 190, 220, 125, 175, 155, 195]

# Histogramme
plt.figure(figsize=(8, 6))
plt.hist(data2, bins=5, edgecolor='black')
plt.title('Histogramme des dépenses mensuelles')
plt.xlabel('Dépenses (euros)')
plt.ylabel('Fréquence')
plt.grid(True, alpha=0.3)
plt.savefig('ex2_histogramme.png')
plt.show()

# Interprétation
print('Forme : Asymétrique, légèrement étirée à droite.')
print('Étendue : De 120 à 220 euros.')
print('Concentration : Pic autour de 140–180 euros.')
```

## Exercice 3 : Scores d'examen (Éducation)

```python
# Données
data3 = [55, 72, 88, 65, 78, 92, 60, 85, 70, 95, 68, 82, 75, 90, 62, 80, 58, 87, 73, 66, 84, 77, 63, 91, 69]

# Histogramme
plt.figure(figsize=(8, 6))
plt.hist(data3, bins=6, edgecolor='black')
plt.title('Histogramme des scores d'examen')
plt.xlabel('Scores (sur 100)')
plt.ylabel('Fréquence')
plt.grid(True, alpha=0.3)
plt.savefig('ex3_histogramme.png')
plt.show()

# Interprétation
print('Forme : Symétrique, en cloche.')
print('Étendue : De 55 à 95 points.')
print('Concentration : Pic autour de 70–85 points.')
```

## Exercice 4 : Concentration de polluants (Environnement)

```python
# Données
data4 = [15, 22, 18, 25, 30, 20, 28, 17, 23, 26, 19, 24, 21, 27, 16, 29, 20, 22]

# Histogramme
plt.figure(figsize=(8, 6))
plt.hist(data4, bins=5, edgecolor='black')
plt.title('Histogramme des concentrations de PM2.5')
plt.xlabel('Concentration (µg/m³)')
plt.ylabel('Fréquence')
plt.grid(True, alpha=0.3)
plt.savefig('ex4_histogramme.png')
plt.show()

# Interprétation
print('Forme : Uniforme, légère concentration centrale.')
print('Étendue : De 15 à 30 µg/m³.')
print('Concentration : Léger pic entre 18 et 24 µg/m³.')
```

## Exercice 5 : Temps de course (Sport)

```python
# Données
data5 = [20.5, 22.0, 19.8, 23.5, 21.2, 24.0, 20.0, 22.8, 21.5, 23.0, 19.5, 24.5, 20.8, 22.2, 21.0, 23.8, 20.2, 22.5, 21.8, 24.2, 19.7, 23.2]

# Histogramme
plt.figure(figsize=(8, 6))
plt.hist(data5, bins=6, edgecolor='black')
plt.title('Histogramme des temps de course')
plt.xlabel('Temps (minutes)')
plt.ylabel('Fréquence')
plt.grid(True, alpha=0.3)
plt.savefig('ex5_histogramme.png')
plt.show()

# Interprétation
print('Forme : Symétrique, légèrement en cloche.')
print('Étendue : De 19.5 à 24.5 minutes.')
print('Concentration : Pic autour de 20.5–23.0 minutes.')
```
