# Histogramme du taux d'alcalinité de l'eau
Ce notebook affiche un histogramme basé sur les mesures du taux d'alcalinité.

```
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

alcalinite = [
    87, 91, 85, 78, 94, 96, 83, 82, 89, 88,
    92, 97, 95, 86, 90, 93, 84, 80, 81, 88,
    90, 91, 92, 85, 84, 89, 90, 93, 94, 95
]

df = pd.DataFrame({'Taux d\'alcalinité (mg/L)': alcalinite})

plt.figure(figsize=(10, 6))
sns.histplot(df, x='Taux d\'alcalinité (mg/L)', bins=5, kde=False, color='skyblue', edgecolor='black')
plt.title("Histogramme du taux d'alcalinité de l'eau")
plt.xlabel("Taux d'alcalinité (mg/L)")
plt.ylabel("Effectif")
plt.grid(True)
plt.tight_layout()
plt.show()
```
