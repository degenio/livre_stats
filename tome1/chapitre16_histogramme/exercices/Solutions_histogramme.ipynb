# Solutions des exercices sur l'Histogramme

Ce notebook génère les histogrammes pour les 5 exercices en utilisant Python avec `matplotlib` et `pandas`, et décrit les distributions observées.

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

## Exercice 1 : Durée de sommeil (Santé)

```python
# Données
# data1 = [6.5, 7.0, 8.0, 6.8, 7.5, 6.2, 8.5, 7.2, 6.0, 7.8, 8.2, 6.7, 7.4, 6.9, 8.0, 7.1, 6.3, 8.3, 7.6, 6.4]
data1 = [6.5, 7.0, 8.0, 6.8, 7.5, 6.2, 8.5, 7.2, 6.0, 7.8, 8.2, 6.7, 7.4, 6.9, 8.0, 7.1, 6.3, 8.3, 7.6, 6.4]

# Histogramme
plt.figure(figsize=(8, 6))
plt.hist(data1, bins=5, edgecolor='black', color='gray')
plt.title('Histogramme de la durée de sommeil')
plt.xlabel('Durée de sommeil (heures)')
plt.ylabel('Fréquence')
plt.grid(True, alpha=0.3)
plt.savefig('ex1_histogramme.png')
plt.show()

# Interprétation
print('Forme : Assez symétrique, légèrement en cloche.')
print('Étendue : De 6.0 à 8.5 heures.')
print('Concentration : Pic autour de 6.5–7.5 heures.')
```

## Exercice 2 : Dépenses mensuelles (Économie)

```python
#Partie 1
import matplotlib.pyplot as plt

# Données
depenses = [120, 150, 200, 180, 130, 170, 210, 140,
            160, 190, 220, 125, 175, 155, 195]

# Création des histogrammes
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Histogramme avec 3 classes
axes[0].hist(depenses, bins=3, edgecolor='black', color='gray')
axes[0].set_title('Histogramme - 3 classes')
axes[0].set_xlabel('Dépenses mensuelles ($)')
axes[0].set_ylabel('Fréquence')

# Histogramme avec 7 classes
axes[1].hist(depenses, bins=7, edgecolor='black', color='lightgray')
axes[1].set_title('Histogramme - 7 classes')
axes[1].set_xlabel('Dépenses mensuelles ($)')
axes[1].set_ylabel('Fréquence')
fig.suptitle('Dépenses mensuelles - Effet du nombre de classes', fontsize=14)
plt.savefig('ex21_histogramme.png')
plt.tight_layout()
plt.show()
```

```python
# 5 classes
data2 = [120, 150, 200, 180, 130, 170, 210, 140, 160, 190, 220, 125, 175, 155, 195]

# Histogramme
plt.figure(figsize=(8, 6))
plt.hist(data2, bins=5, edgecolor='black', color='gray')
plt.title('Histogramme des dépenses mensuelles')
plt.xlabel('Dépenses ($)')
plt.ylabel('Fréquence')
plt.grid(True, alpha=0.3)
plt.savefig('ex2_histogramme.png')
plt.show()

# Interprétation
print('Forme : La distribution est uniforme.')
print('Étendue : De 120 à 220 $.')
print('Concentration : Les valeurs sont uniformément réparties sur les 5 classes, avec une fréquence constante de 3 ménages par classe.')
```

## Exercice 6 : Histogrammes de prédicteurs pour un modele de regression

```python
import matplotlib.pyplot as plt

# Données
surface = [60, 85, 70, 120, 55, 90, 65, 110, 75, 100, 50, 95, 80, 130, 70]
nb_pieces = [2, 3, 2, 4, 1, 3, 2, 4, 2, 3, 1, 3, 2, 4, 2]
prix = [250000, 320000, 280000, 450000, 200000, 350000, 260000,
        420000, 290000, 380000, 190000, 360000, 310000, 480000, 270000]

# Créer une figure avec 3 sous-graphiques
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Histogramme - Surface
axes[0].hist(surface, bins=5, color='gray', edgecolor='black')
axes[0].set_title('Surface (m²)')
axes[0].set_xlabel('Surface (m²)')
axes[0].set_ylabel('Fréquence')

# Histogramme - Nombre de Pièces
axes[1].hist(nb_pieces, bins=4, color='lightgray', edgecolor='black', 
             range=(0.5, 4.5))  # range pour centrer les barres sur les entiers
axes[1].set_title('Nombre de Pièces')
axes[1].set_xlabel('Nombre de Pièces')
axes[1].set_xticks([1, 2, 3, 4])
axes[1].set_ylabel('Fréquence')

# Histogramme - Prix
axes[2].hist(prix, bins=5, color='gray', edgecolor='black')
# axes[2].set_title('Prix ($)')
axes[2].set_xlabel('Prix ($)')
axes[2].set_ylabel('Fréquence')
plt.savefig('ex6_histogramme.png')
# Ajuster l'espacement
plt.tight_layout()
plt.show()
```

## Exercice 6 : Histogrammes de prédicteurs pour un modele de regression

```python
import matplotlib.pyplot as plt

# Données de température
temperatures = [
    18.2, 19.5, 20.1, 18.8, 21.3, 19.9, 20.5, 18.5,
    19.2, 20.8, 21.0, 19.6, 20.3, 18.9, 21.2, 19.4,
    20.6, 18.7, 19.8, 20.9, 21.1, 19.1, 20.4, 18.6,
    19.7, 20.2, 21.4, 19.3
]

# Nombre de classes selon Sturges
import math
n = len(temperatures)
k = round(1 + math.log2(n))  # k = 6
print(k)
# Tracer l'histogramme
plt.figure(figsize=(8, 5))
plt.hist(temperatures, bins=k, color='gray', edgecolor='black')
plt.title(f'Histogramme des températures sur {n} jours\n({k} classes selon la règle de Sturges)')
plt.xlabel('Température (°C)')
plt.ylabel('Fréquence')
# plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.savefig('ex7_histogramme.png')
plt.tight_layout()
plt.show()
```
