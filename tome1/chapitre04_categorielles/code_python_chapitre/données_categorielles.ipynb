# Les exemples suivants en python comment faire un encodage nominal et un encodage ordinal.

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, OneHotEncoder

# Données exemple
df = pd.DataFrame({
    'couleur': ['rouge', 'bleu', 'vert', 'rouge', 'bleu'],
    'satisfaction': ['Faible', 'Moyen', 'Élevé', 'Moyen', 'Faible'],
    'region': ['Nord', 'Sud', 'Nord', 'Est', 'Ouest']
})

# Encodage ordinal (préserve l'ordre)
satisfaction_mapping = {'Faible': 1, 'Moyen': 2, 'Élevé': 3}
df['satisfaction_encoded'] = df['satisfaction'].map(satisfaction_mapping)

# One-hot encoding (nominal)
df_encoded = pd.get_dummies(df, columns=['couleur', 'region'], prefix=['couleur', 'region'])

print("Données encodées :")
df_encoded.head()
```
