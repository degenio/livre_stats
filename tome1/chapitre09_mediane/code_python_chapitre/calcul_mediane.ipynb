# Code de calcul de la médiane dans différentes situations.

```python
import numpy as np
import pandas as pd

# Données exemple (temps de réponse)
temps = np.array([12, 18, 15, 20, 28, 25, 22, 60, 32, 35, 
                  38, 40, 45, 52, 50, 48, 55, 90, 30, 62, 
                  68, 65, 70, 72, 75, 78, 80, 82, 85, 88])

# Médiane avec numpy
mediane = np.median(temps)
print(f"Médiane : {mediane:.2f} secondes")

# Comparaison moyenne vs médiane avec outlier
donnees_avec_outlier = np.array([12, 15, 18, 20, 22, 1000])  # 1000 = outlier
print(f"\nAvec outlier (1000) :")
print(f"  Moyenne : {np.mean(donnees_avec_outlier):.2f}")
print(f"  Médiane : {np.median(donnees_avec_outlier):.2f}")

# Vérification : médiane = Q2 = percentile 50
print(f"\nVérification : médiane = Q2 = P50")
print(f"  np.percentile(temps, 50) = {np.percentile(temps, 50):.2f}")
print(f"  np.median(temps) = {np.median(temps):.2f}")

# Avec pandas
s = pd.Series(temps)
print(f"\nAvec pandas : médiane = {s.median():.2f}")
```
