# Code de calcul des quantiles dans différentes situations.

```python
import numpy as np
import pandas as pd

# Données exemple (notes triées)
notes = np.array([28, 32, 35, 40, 42, 45, 48, 50, 52, 55, 
                  57, 60, 62, 65, 68, 70, 72, 75, 78, 82, 
                  85, 88, 90, 94, 100])

# Calcul des percentiles avec différentes méthodes
print("Méthode 'linear' (numpy par défaut) :")
print(f"P25 = {np.percentile(notes, 25, method='linear'):.2f}")
print(f"P50 = {np.percentile(notes, 50, method='linear'):.2f}")
print(f"P75 = {np.percentile(notes, 75, method='linear'):.2f}")
 
# Avec pandas (utilise numpy en backend)
s = pd.Series(notes)
print(f"\nAvec pandas : Q1={s.quantile(0.25):.2f}, Médiane={s.quantile(0.5):.2f}, Q3={s.quantile(0.75):.2f}")

# Comparaison rapide des méthodes pour P25
methods = ['linear', 'lower', 'higher', 'midpoint', 'nearest']
print(f"\nComparaison des méthodes pour P25 :")
for m in methods:
    val = np.percentile(notes, 25, method=m)
    print(f"  {m:10s} : {val:.2f}")
```
